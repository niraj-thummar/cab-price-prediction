```python
import pandas as pd
```

```python
df = pd.read_csv('rideshare_kaggle.csv');
```

```python
df['price'].isnull().sum
```

```python
df.dropna(subset=['price'], inplace=True)
```

```python
df['price'].isnull().sum()
```

```python
df.shape
```

```python
columns_to_keep = [
    'cab_type', 'name', 'distance', 'surge_multiplier', 
    'source', 'destination', 'hour', 'day', 'month', 
    'temperature', 'short_summary', 'precipIntensity', 'price'
]

df = df[columns_to_keep]

```

```python
df.shape
```

```python
df.columns
```

```python
df = df[df['price'] <= 45]
#remove outliers because >45$ prices are outliers 
```

```python
df.shape
```

```python
 df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
Q1 = df['distance'].quantile(0.25)
Q3 = df['distance'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['distance'] < lower) | (df['distance'] > upper)]

print(outliers)
print("Number of outliers:", len(outliers))
```

```python
df = df[(df['distance'] >= lower) & (df['distance'] <= upper)]
```

```python
df.shape
```

```python
categorical_cols = ['cab_type', 'name', 'source', 'destination', 'short_summary']
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
df_encoded.shape
```

```python
df.shape
```

```python
df_encoded.columns
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
numeric_cols = ['distance', 'surge_multiplier', 'hour', 'day', 'month', 'temperature', 'precipIntensity']

scaler = MinMaxScaler()
```

```python
df_encoded[numeric_cols] = scaler.fit_transform(df_encoded[numeric_cols])
```
